# Prediction: exact-solution plug-in estimators and direct transformers

Both seeds; M=2^19; training n=128; the same first 256 test examples per seed at each inference n. Existing MLE/SGD estimates and transformer caches are loaded; no fitting or network inference is run.

Parameter transformer: decode its mean parameter output and apply the exact trajectory at the saved query time. MLE/SGD: apply that same trajectory to their saved estimates. Direct prediction transformer: take its cached mean token output and calculate its squared error, averaged over state coordinates (L_mean). Joint model: plot both its cached mean state output directly and its decoded mean parameter output composed with the exact solution, for training n=128. Joint caches are checksum-verified and checked against the same test parameters, query times, and targets. Thus every method returns one state estimate per example. All errors use the same state normalization and clean target x(tau), including birth–death. Parameters are not clipped before composition.

Bands are approximate pointwise 95% percentile-bootstrap intervals for mean test error, using 5,000 resamples of the 256 whole test trajectories, separately for each seed. These intervals are nonnegative; a zero lower endpoint is omitted on the logarithmic axis instead of clipped to an arbitrary value. Bootstrap intervals can still be unreliable when rare extreme errors dominate this small test bank. Each curve averages both seed-specific risks equally. Tau covers the full prediction interval. For forced linear, the saved posterior mean parameter estimates are also composed with the exact solution (equivalently, the posterior mean trajectory for this affine family). Only matching, converged caches are loaded; no posterior integration is rerun. Other families omit the posterior curve. Figures save automatically to figures/prediction_statistical_comparisons/.

Plots use equal-weight averages across the two trained seeds (errors, not predictions). Test-error bands are within-seed bootstrap intervals for this average, conditional on the trained models, not across-training-seed confidence intervals. Original figure filenames are retained.

## Side-by-side: parameter estimation and trajectory prediction

For each family, the left panel repeats the parameter-estimation comparison (notebook 03) and the right panel the prediction comparison above. Curves and bands are read from the saved `*_summary.json` files, so no fitting or inference is rerun; run notebook 03 and the cells above first if those summaries are stale. Each estimator has the same color and marker in both panels; in the prediction panel, parameter estimators are composed with the exact ODE solution. Figures are saved under `figures/statistical_comparisons_side_by_side/`.

```python
# Side-by-side statistical comparison: parameter estimation (left) and trajectory prediction (right).
# Reads the saved curve summaries only (no fitting or inference):
#   parameter panel  <- figures/statistical_comparisons/{family}_summary.json   (written by notebook 03)
#   prediction panel <- figures/prediction_statistical_comparisons/{family}_summary.json   (written above)
import json
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.transforms import Bbox
importlib.reload(paper_style)
from ode_paper.paper_style import apply_style, BAND_ALPHA, context_ticks

LEGEND_GAP = 0.06  # inches between the x-axis labels and the top of the legend
FAMILY_ABBR = {'forced_linear': 'FL', 'logistic': 'LG', 'birth_death': 'BD', 'stuart_landau': 'SL'}
# One style per estimator, shared by both panels. In the prediction panel, parameter
# estimators are composed with the exact ODE solution.
ESTIMATORS = {
    'param_tf':   dict(label='Parameter-only transformer',     color='C0', marker='o'),
    'pred_tf':    dict(label='Prediction-only transformer',    color='C1', marker='^'),
    'joint_par':  dict(label='Joint transformer (parameters)', color='C2', marker='o'),
    'joint_pred': dict(label='Joint transformer (prediction)', color='C3', marker='^'),
    'mle':        dict(label='Bounded MLE',                    color='C4', marker='*'),
    'sgd':        dict(label='Projected SGD',                  color='C5', marker='*'),
    'posterior':  dict(label='Posterior mean',                 color='C6', marker='*'),
}
PARAMETER_METHODS = {
    'Transformer train n=128': 'param_tf',
    'Joint parameter train n=128': 'joint_par',
    'Bounded MLE': 'mle', 'Projected SGD': 'sgd', 'Posterior mean': 'posterior',
}
PREDICTION_METHODS = {
    'Parameter transformer + solution (train n=128)': 'param_tf',
    'Prediction transformer (train n=128)': 'pred_tf',
    'Joint parameter + solution (train n=128)': 'joint_par',
    'Joint direct prediction (train n=128)': 'joint_pred',
    'Bounded MLE + solution': 'mle', 'Projected SGD + solution': 'sgd',
    'Posterior mean + solution': 'posterior',
}

def _curves(path, method_keys):
    data = json.loads(path.read_text())
    rows = data['curves'] if isinstance(data, dict) else data
    curves = {}
    for method, key in method_keys.items():
        pts = sorted((r for r in rows if r['method'] == method), key=lambda r: r['n'])
        if pts:
            curves[key] = {k: np.array([r[f] for r in pts], dtype=float)
                           for k, f in (('n', 'n'), ('mean', 'mean'), ('low', 'ci95_lower'), ('high', 'ci95_upper'))}
    if not curves:
        raise ValueError(f'No recognized curves in {path}')
    return curves

def _panel(ax, curves, title, ylabel):
    for key, style in ESTIMATORS.items():
        if key not in curves:
            continue
        c = curves[key]
        star = style['marker'] == '*'
        ax.plot(c['n'], c['mean'], color=style['color'], marker=style['marker'],
                markersize=7 if star else 4, linewidth=2.1, zorder=3)
        valid = c['low'] > 0  # zero lower endpoints are omitted on the log axis, as in the single plots
        ax.fill_between(c['n'], np.where(valid, c['low'], np.nan), c['high'], where=valid,
                        color=style['color'], alpha=BAND_ALPHA, linewidth=0, zorder=1)
    ns = next(iter(curves.values()))['n'].astype(int)
    ax.set(xscale='log', yscale='log', xlabel='Inference observations n', ylabel=ylabel, title=title)
    context_ticks(ax, ns)
    ax.grid(True, which='both', alpha=.18)

def plot_side_by_side(family):
    apply_style()
    param = _curves(ROOT/'figures'/'statistical_comparisons'/f'{family}_summary.json', PARAMETER_METHODS)
    pred = _curves(ROOT/'figures'/'prediction_statistical_comparisons'/f'{family}_summary.json', PREDICTION_METHODS)
    abbr = FAMILY_ABBR.get(family, family)
    fig, (ax_par, ax_pred) = plt.subplots(1, 2, figsize=(8.0, 3.0))
    _panel(ax_par, param, f'Parameter estimation ({abbr})',
           'MSE of (b-d,c)' if family == 'birth_death' else 'Normalized MSE')
    _panel(ax_pred, pred, f'Trajectory prediction ({abbr})', 'Normalized MSE')
    # Same canvas and axes layout as the paired plots in notebook 02 (save_pair): 8.0 x 3.0 in
    # figure with the axes laid out above the bottom 16%, and one shared legend below.
    if ax_pred.get_ylabel() == ax_par.get_ylabel():
        ax_pred.set_ylabel('')
    shown = [k for k in ESTIMATORS if k in param or k in pred]
    handles = [Line2D([], [], color=ESTIMATORS[k]['color'], marker=ESTIMATORS[k]['marker'],
                      markersize=7 if ESTIMATORS[k]['marker'] == '*' else 4, linewidth=2.1) for k in shown]
    one_row = len(shown) <= 5
    legend = fig.legend(handles, [ESTIMATORS[k]['label'] for k in shown], loc='upper center',
                        bbox_to_anchor=(.5, .16), ncol=len(shown) if one_row else 4,
                        fontsize=10 if one_row else 8, handlelength=3 if one_row else 1.8,
                        columnspacing=.8, handletextpad=.4, framealpha=.95)
    fig.tight_layout(rect=(0, .16, 1, 1))
    # Place the legend just below the x-axis labels (axes boxes keep the notebook-02 size).
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    to_fig = fig.transFigure.inverted()
    label_bottom = min(to_fig.transform(ax.get_tightbbox(renderer))[0][1] for ax in fig.axes)
    legend.set_bbox_to_anchor((.5, label_bottom - LEGEND_GAP/fig.get_figheight()), transform=fig.transFigure)
    fig.canvas.draw()
    legend_bottom = legend.get_window_extent(renderer).y0/fig.dpi
    output = ROOT/'figures'/'statistical_comparisons_side_by_side'
    output.mkdir(parents=True, exist_ok=True)
    for ext in ('png', 'pdf'):
        # Crop only the empty strip below the legend; the full 8 in width is kept so the
        # panels scale to the same size as the other paired figures at text width.
        fig.savefig(output/f'{family}.{ext}', dpi=180,
                    bbox_inches=Bbox([[0, max(legend_bottom-.04, 0)], [fig.get_figwidth(), fig.get_figheight()]]))
    plt.show()

for family in ('forced_linear', 'logistic', 'birth_death', 'stuart_landau'):
    plot_side_by_side(family)
```

```python
# First prediction comparison, retaining only the four transformer curves.
# Uses the saved seed-averaged means and confidence intervals from plot 1.
import json
import numpy as np
import matplotlib.pyplot as plt
from ode_paper.paper_style import apply_style, BAND_ALPHA, context_ticks
from ode_paper.prediction_comparison import PREDICTION_LABELS

apply_style()
summary_path = ROOT / 'figures' / 'prediction_statistical_comparisons' / 'forced_linear_summary.json'
saved = json.loads(summary_path.read_text())
saved = saved['curves'] if isinstance(saved, dict) else saved
all_methods = list(dict.fromkeys(r['method'] for r in saved))
model_names = ('Parameter transformer + solution (train n=128)',
               'Prediction transformer (train n=128)',
               'Joint parameter + solution (train n=128)',
               'Joint direct prediction (train n=128)')
model_methods = [m for m in all_methods if m in model_names]
if len(model_methods) != 4:
    raise ValueError(f'Expected four transformer curves, found {model_methods}')
fig, ax = plt.subplots(figsize=(8.0, 3.2))
for method in model_methods:
    points = sorted((r for r in saved if r['method'] == method), key=lambda r: r['n'])
    ns, mean, low, high = (np.array([r[k] for r in points]) for k in
                          ('n', 'mean', 'ci95_lower', 'ci95_upper'))
    color = f'C{all_methods.index(method)}'
    direct = 'direct prediction' in method.lower() or method.startswith('Prediction transformer')
    label = PREDICTION_LABELS.get(method.replace(' (train n=128)', ''),
                                  method.replace(' (train n=128)', ''))
    ax.plot(ns, mean, color=color, marker='^' if direct else 'o',
            markersize=4, linewidth=2.1, label=label, zorder=3)
    valid = low > 0
    ax.fill_between(ns, np.where(valid, low, np.nan), high, where=valid,
                    color=color, alpha=BAND_ALPHA, linewidth=0, zorder=1)
ax.set(xscale='log', yscale='log', xlabel='Inference observations n',
       ylabel='Normalized MSE', title='Prediction model comparison (FL)')
context_ticks(ax, ns.astype(int))
ax.grid(True, which='both', alpha=.15)
fig.tight_layout()
ax.legend(loc='upper center', bbox_to_anchor=(.5, -.27), ncol=4, fontsize=8,
          handlelength=1.4, columnspacing=.8, handletextpad=.4,
          framealpha=.95, borderaxespad=0)
output = ROOT / 'figures' / 'prediction_statistical_comparisons'
for ext in ('png', 'pdf'):
    fig.savefig(output / f'forced_linear_models_only.{ext}', dpi=180, bbox_inches='tight')
plt.show()

```
